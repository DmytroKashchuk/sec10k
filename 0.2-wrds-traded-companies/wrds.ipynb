# WRDS

```python
#!/usr/bin/env python3
"""
crsp_public_companies.py
------------------------
Pull the universe of US publicly traded companies from CRSP via WRDS
and save it as CSV.

Output: crsp_public_companies.csv
"""

import wrds

db = wrds.Connection()

# active US common stock on NYSE / AMEX / Nasdaq
df = db.raw_sql("""
    SELECT DISTINCT *
    FROM crsp.stocknames
""")

# decode exchange code into a readable label
exchange_map = {1: "NYSE", 2: "AMEX", 3: "Nasdaq"}
df["exchange"] = df["exchcd"].map(exchange_map)

print(f"{len(df):,} companies")
df.to_csv("crsp_public_companies.csv", index=False)
print("wrote crsp_public_companies.csv")

db.close()
```
